```python
import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path

import torch
import torch.nn as nn
import torchvision
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset, random_split, Subset
from torchvision import datasets
import torchvision.transforms as transforms
from PIL import Image, ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES=True
from collections import Counter

from sklearn.metrics import accuracy_score, f1_score, recall_score, precision_score, confusion_matrix, classification_report
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

from skimage.feature import graycoprops, graycomatrix, hog

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
```

```python
img_size=224

DATA_DIR = Path('data')

dataset =  datasets.ImageFolder(root="Data")

# split
# way - 1
# train_size = int(0.7 * len(dataset))
# val_size = int(0.15 * len(dataset))
# test_size = len(dataset) - train_size - val_size

# train_set, val_set, test_set = random_split(dataset, [train_size, val_size, test_size])

# train_set.dataset.transform = train_transform
# test_set.dataset.transform = test_transform

# way - 2

n = len(dataset)

train_size = int(0.7 * n)
val_size = int(0.15 * n)
test_size = n - train_size - val_size

indices = torch.randperm(n)
train_idx = indices[:train_size]
val_idx = indices[train_size:train_size + val_size]
test_idx = indices[train_size + val_size:]

train_transform = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomCrop(img_size),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

test_transform = transforms.Compose([
    transforms.Resize((img_size, img_size)),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

train_dataset = datasets.ImageFolder(root="Data", transform=train_transform)
eval_dataset = datasets.ImageFolder(root="Data", transform=test_transform)
test_dataset = datasets.ImageFolder(root="Data", transform=test_transform)

trainset = Subset(train_dataset, train_idx)
valset = Subset(eval_dataset, val_idx)
testset = Subset(test_dataset, test_idx)
```

```python
trainloader = DataLoader(trainset, batch_size=64, shuffle=True)
valloader = DataLoader(valset, batch_size=64)
testloader = DataLoader(testset, batch_size=64)
```

```python
classes = sorted([f.name for f in DATA_DIR.iterdir() if f.is_dir()])

records = []

for cls in classes:
    cls_dir = DATA_DIR/cls
    for fname in cls_dir.iterdir():
        if fname.suffix.lower() in (".jpg", ".jpeg", ".png"):
            records.append({
                "filepath": str(fname),
                "label": cls,
                "filename": fname.name
            })

df = pd.DataFrame(records)
df.head()
```

```python
class_counts = df["label"].value_counts()
print(class_counts)

plt.figure(figsize=(7,5))
ax = sns.barplot(x=class_counts.index, y=class_counts.values, palette="viridis")
plt.title("Class Distribution - Rice Leaf Disease Dataset")
plt.ylabel("Number of Images")
plt.xlabel("Disease Class")
for i, v in enumerate(class_counts.values):
    ax.text(i, v + 0.3, str(v), ha="center", fontweight="bold")
plt.tight_layout()
Path("outputs_figs").mkdir(parents=True, exist_ok=True)
plt.savefig("outputs_figs/class_distribution.png", dpi=100)
plt.show()
```

```python
def read_image_uint8(filepath, size=None):
    img = Image.open(filepath).convert("RGB")

    if size is not None:
        img = img.resize(size)
    return np.array(img)

classes = df['label'].unique()

fig, axes = plt.subplots(len(classes), 5, figsize=(18,11))

for i, cls in enumerate(classes):
    sample_files = (
        df[df['label']==cls]['filepath'].sample(5, random_state=42).tolist()
    )

    for j, fp in enumerate(sample_files):
        img = read_image_uint8(fp)

        axes[i, j].imshow(img)
        axes[i, j].axis("off")

    axes[i, 0].set_title(
        cls, fontsize=13, fontweight="bold", loc="left"
    )

plt.suptitle("Sample Images per Class", fontsize=16)

plt.tight_layout()
plt.show()
```
